Converting color image to grey image

```python
import cv2
print('Original Image')
image=cv2.imread('/content/sun.jpg')
from google.colab.patches import cv2_imshow
cv2_imshow(image)
print('Grey_image')
grey_image=cv2.imread('/content/sun.jpg',0)
cv2_imshow(grey_image)
```

Converting image to array number of image

```python
from numpy import asarray
data=asarray(image)
print(data)
```

Image Rotation

```python
import imutils
rotated=imutils.rotate(image,45)
cv2_imshow(rotated)
print('Rotated image')
```

Conversion of RGB Model to CMY Model

```python
print('RGB image')
R,G,B=cv2.split(image)
print('R component')
cv2_imshow(R)
print('G component')
cv2_imshow(G)
print('B component')
cv2_imshow(B)

print('CMY image')
C=1-R
M=1-G
Y=1-B
print('C component')
cv2_imshow(C)
print('M component')
cv2_imshow(M)
print('Y component')
cv2_imshow(Y)

```
