mapping income into binary

```python
if df['income'].dtype == 'object':
    df['income'] = df['income'].str.strip()
    print(df['income'].unique())
    df['income'] = df['income'].map({'<=50K':0, '>50K':1}).astype(int)
df.head()
```

```python
pd.DataFrame({'null':df.isnull().sum()})
```

segretion of dataset (divided into x and y

```python
x=df.iloc[:,:-1]#divided in input,first all row will come and then all col will comme except the last one
y=df.iloc[:,-1]#divided in output,all row will come aand column just last one

x.head()
```

```python
y.head()
```

```python
x.tail()
```

```python
y.tail()
```

split( x and y both will  divided into two parts

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.25,random_state=0)
```

feature scaling

```python
# 1. Encode categorical data
x_train = pd.get_dummies(x_train)
x_test = pd.get_dummies(x_test)
# 2. Apply scaling
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

finding k-value

```python
error=[]
from sklearn.neighbors import KNeighborsClassifier
for i in range (1,40):
  model=KNeighborsClassifier(n_neighbors=i)
  model.fit(x_train,y_train)
  pred_i=model.predict(x_test)
  error.append(np.mean(pred_i !=y_test))
plt.figure(figsize=(12,6))
plt.plot(range(1,40),error,color='red',marker='o',markerfacecolor='blue', linestyle='dashed')
plt.title('error rate k-value')
plt.xlabel('k-value')
plt.ylabel('Mean error')
```

\Train and predict data

```python
model=KNeighborsClassifier(n_neighbors=36,metric='minkowski',p=2)
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
```

checking accuracy/validation/evaluation

```python
from sklearn.metrics import confusion_matrix,accuracy_score
print('Accuracy model:{0}%'.format(accuracy_score(y_test,y_pred)*100))
```

PyTorch dataset creation

```python
from transformers import AutoModelForSequenceClassification
model=AutoModelForSequenceClassification.from_pretrained(
    'bert-base-uncased',
    num_labels=2
)
```

Train Model

```python
from transformers import Trainer, TrainingArguments

training_args = TrainingArguments(
    output_dir='./results',
    max_steps=10,
    report_to='none'

)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=dataset
)

trainer.train()
```
